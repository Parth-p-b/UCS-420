## Q6

```python
employees_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(employees_data)
employees["Joining_Date"] = pd.to_datetime(employees["Joining_Date"])

employees.to_csv("employees.csv", index=False)
employees
```

```python
print("Shape:", employees.shape)
```

```python
employees.info()
```

```python
employees.describe(include='all')
```

```python
print("First 5 rows:")
display(employees.head(5))

print("Last 3 rows:")
display(employees.tail(3))
```

```python
print("Average salary:", employees["Salary"].mean())
print("Total bonus:", employees["Bonus"].sum())
print("Youngest employee's age:", employees["Age"].min())
print("Highest performance rating:", employees["Rating"].max())
```

```python
employees_sorted = employees.sort_values(by="Salary", ascending=False)
employees_sorted
```

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance_Category"] = employees["Rating"].apply(performance_category)
employees
```

```python
print(employees.isnull().sum())
```

```python
employees.rename(columns={"Employee_ID": "ID"}, inplace=True)
employees
```

```python
print("Employees with more than 5 years of experience:")
display(employees[employees["Years_of_Experience"] > 5])

print("Employees in the IT department:")
display(employees[employees["Department"] == "IT"])
```

```python
employees["Tax"] = employees["Salary"] * 0.10
employees
```

```python
employees.to_csv("employees_modified.csv", index=False)
print("Saved as employees_modified.csv")
```
